### Задание 1.
Определить, пересекаются ли плоскости (все сразу). Если пересекаются, найти точку пересечения.

Составить СЛАУ в виде списка уравнений, решить с помощью linsolve.

$-8x+5y-2z+28=0$

$-11x-10y-16z+71=0$

$-7x+3y+8z-27=0$.

```python
x, y, z = symbols('x y z')
eq1 = -8*x + 5*y - 2*z + 28
eq2 = -11*x - 10*y - 16*z + 71
eq3 = -7*x + 3*y + 8*z - 27
equations = [eq1, eq2, eq3]
result = linsolve(equations, x, y, z)
if result:
    print(f"Плоскости пересекаются в точке с координатами: {list(result)[0]}")
else:
    print("Плоскости не пересекаются.")
```

```python
t1 = symbols('t1')
a1 = B - A
X = Matrix([x, y, z])
AB = A + t1 * a1
AB_sys = [Eq(X[i], AB[i]) for i in range(len(AB))]
display(*AB_sys)
```

```python
XK = list(X - K)
MK = list(M - K)
NK = list(N - K)
n = Matrix([XK, MK, NK]).det()
display(n)
```
