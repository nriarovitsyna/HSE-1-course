```python
A = Matrix([[0, -4, 0, 1], [1, -4, -4, 2], [4, 1, 4, -3], [1, -8, -4, 3], [0, -4, 0, 1]])
b1 = Matrix([-3, -1, 2, -4, -3])
b2 = Matrix([-3, 1, 0, 4, 0])
Ab1_rank = A.row_join(b1).rank()
Ab2_rank = A.row_join(b2).rank()
A_rank = A.rank()
print(f'Ранг {A_rank} матрицы A{" не" * (A_rank != Ab1_rank)} равен рангу {Ab1_rank} расширенной матрицы => СЛАУ {"не" * (A_rank != Ab1_rank)}совместна')
print(f'Ранг {A_rank} матрицы A{" не" * (A_rank != Ab2_rank)} равен рангу {Ab2_rank} расширенной матрицы => СЛАУ {"не" * (A_rank != Ab2_rank)}совместна')
```

```python
B = Matrix([[3, 0, -4, -4], [2, -3, 4, -1], [-2, 0, 2, -2], [1, -1, -2, 3]])
b1 = Matrix([3, -1, -2, 0])
b2 = Matrix([0, -1, -2, 0])
Bb1_rank = B.row_join(b1).rank()
Bb2_rank = B.row_join(b2).rank()
B_rank = B.rank()
print(f'Ранг {B_rank} матрицы B{" не" * (B_rank != Bb1_rank)} равен рангу {Bb1_rank} расширенной матрицы => СЛАУ {"не" * (B_rank != Bb1_rank)}совместна')
print(f'Ранг {B_rank} матрицы B{" не" * (B_rank != Bb2_rank)} равен рангу {Bb2_rank} расширенной матрицы => СЛАУ {"не" * (B_rank != Bb2_rank)}совместна')
```

Для соответствующей однородной СЛАУ проверить существование нетривиального решения. В случае, если оно существует,
найти размерность пространства решений и составить ФСР и общее решение однородной и неоднородной СЛАУ.


```python
A = Matrix([[4, 7, -1, -5, 7], [-2, 5, 0, -9, -6], [8, 48, -4, -56, 4], [24, 8, -4, 16, 52],[-4, 0, -3, 5, -3]])
b = Matrix([-9, -77, -344, 272, -13])
A_rank = A.rank()
Ab_rank = A.row_join(b).rank()
print(f'Ранг {A_rank} матрицы A{" не" * (A_rank != Ab_rank)} равен рангу {Ab_rank} расширенной матрицы => СЛАУ {"не" * (A_rank != Ab_rank)}совместна')
```

```python
print(f'pазмерность подпространства решений = {m - A_rank}')
```

```python
def f(A, x = Symbol('x')):
  A_fsr = A.nullspace()
  X = zeros(A.shape[1], 1)
  for i, vect in enumerate(A_fsr):
    X += vect * Symbol('alpha' + str(i+1))
  return A_fsr, X
display(Latex(f'ФСР\ однородной\ СЛАУ: {latex(f(A)[0])},\ общее\ решение\ СЛАУ: {latex(f(A)[1].T)}'))
Ab = A.row_join(b)
display(Latex(f'ФСР\ неоднородной\ СЛАУ: {latex(f(Ab)[0])},\ общее\ решение\ СЛАУ: {latex(f(Ab)[1].T)}'))
```
