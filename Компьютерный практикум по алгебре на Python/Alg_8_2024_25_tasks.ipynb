### Задание 1.
Решить СЛАУ с помощью linalg.solve() и выполнить проверку подстановкой.
$$
\left\{
\begin{matrix}
-x+4y+3z=18\\
5x-y+4z=81\\
-3x+5y+5z=21
\end{matrix}
\right.
$$

```python
A = np.array([[-1, 4, 3], [5, -1, 4], [-3, 5, 5]])
b = np.array([18, 81, 21])
X = linalg.solve(A, b)
print(f'Решение СЛАУ {X}')
```

```python
np.allclose(A @ X, b)
```

4) расширенную матрицу, полученную из матрицы 3) и столбца 2)

5) созданную на основе списочного выражения np.array $3\times 4$ из расположенных в шахматном порядке чисел $a$ и $-a$, где $a$ - правый нижний элемент матрицы 3).

Применить полученную функцию к расширенным матрицам СЛАУ из Задания 2 со значением **end="\\n\\n"**, а затем  **end** по умолчанию.

```python
def print_info(Ab, end = "\n"):
  A1 = Ab[:,range(0,3)]
  print("1) Матрицa левой части:")
  print(A1, end=end)
  b1 = Ab[:,-1]
  print("2) Столбец правой части:")
  print(b1, end=end)
  A1 = np.transpose(A1)
  print("3) Транспонированная матрица, полученная в пункте 1):")
  print(A1, end=end)
  Ab_1 = np.column_stack((A1, b1))
  print("4) Расширенная матрица из матрицы 3) и столбца 2):")
  print(Ab_1, end=end)
  A = np.array([[A1[2][2] if (i + j) % 2 == 0 else A1[2][2] * -1 for i in range(4)]for j in range(3)])
  print("5) np.array в шахматном порядке:")
  print(A, end= end)
```

### Задание 4.
Исследовать на совместность СЛАУ с параметром $a$ при значениях параметра $a=-1$, $a=0$ и $a=1$ и найти решение, если оно единственно и провести проверку подстановкой.
$$
\left\{
\begin{matrix}
-x+5y-3z=8a\\
4x-ay+5z=-a\\
3x+4y+2z=5a
\end{matrix}
\right.
$$
**Указание** - описать функцию check_SLAE от аргументов $A$ и $b$ (матрица левой части и столбец правой части), которая возвращает 0 для несовместной СЛАУ, 1 для совместной СЛАУ с единственным решением и 2 в остальных случаях.

Затем в цикле по значениям параметра $a$ проверять с помощью check_SLAE совместность и единственность решения СЛАУ и выводить на экран значение параметра и решение СЛАУ, если оно существует и единственно или "решений нет" или "решение не единственно".

```python
def check_SLAE(A, b):
  matrix = np.column_stack((A, b))
  rng_A = np.linalg.matrix_rank(A)
  rng_matrix = np.linalg.matrix_rank(matrix)
  if (rng_A == rng_matrix) and (rng_matrix == A.shape[0]):
    res = linalg.solve(A, b)
    print(f"Решение СЛАУ:")
    display(res)
    return 1
  elif  (rng_A == rng_matrix) and (rng_matrix != A.shape[0]):
    print("Решение не единственно")
    return 2
  else:
    print("Решений нет")
    return 0
```

```python
params = {-1: np.array([[-1, 5, -3],
                        [4, 1, 5],
                        [3, 4, 2]]),
          0: np.array([[-1, 5, -3],
                      [4, 0, 5],
                      [3, 4, 2]]),
          1: np.array([[-1, 5, -3],
                      [4, -1, 5],
                      [3, 4, 2]])}
```

```python
for a in params:
    A = params[a]
    b = np.array([8 * a, -1 * a, 5 * a])
    print(f'При a = {a}:')
    check_SLAE(A, b)
```
