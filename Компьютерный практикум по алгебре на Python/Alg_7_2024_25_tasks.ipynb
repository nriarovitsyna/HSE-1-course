### Задание 1.
Даны точки: $A(7, 5, 4)$,  $B(6, 1, 6)$, $C(5, -3, 8)$, $D(2, 2, 6)$, $F(-7, 8, -1)$.
Составить список точек и словарь точек (ключи - точки, значения - имена точек).

Составить, пользуясь списком и словарем точек, список и словарь прямых (ключи - прямые Line, значения - имена прямых, состоящие из имен точек, например AD). Рассматривать только прямые, проведенные по различным парам точек, например, $AB$ и $AC$ это разные прямые, а $AB$ и $BA$ это одна и та же прямая, в словарь включаем только $AB$, не включаем $BA$.

Вывести на экран значения записей словаря (т.е. имена прямых).

Найти среди прямых словаря пары совпадающих прямых, вывести на экран их имена.

```python
A = Point(7, 5, 4)
B = Point(6, 1, 6)
C = Point(5, -3, 8)
D = Point(2, 2, 6)
F = Point(-7, 8, -1)
points_list = [A, B, C, D, F]
```

```python
points_dict = {
    A: 'A',
    B: 'B',
    C: 'C',
    D: 'D',
    F: 'F'
}
```

```python
from itertools import combinations
lines_list = list()
lines_dict = dict()
for point1, point2 in combinations(points_dict.keys(), 2):
    line_key = Line(point1, point2)
    lines_list.append(line_key)
    lines_dict[line_key] = f"{points_dict[point1]}{points_dict[point2]}"
for line in lines_dict.keys():
    display(Latex(f'{latex(lines_dict[line])}'))
```

```python
for i in range (len(lines_list)):
  for j in range (i + 1, len(lines_list)):
    line1 = Line(lines_list[i])
    line2 = Line(lines_list[j])
    if line1.equals(line2):
      display(Latex(f'{latex(lines_dict[line1])}\ и\ {latex(lines_dict[line2])}'))
```

###  Задание 2
Составить список всех плоскостей, проходящих через любые три точки Задания 1 (плоксости, построенные по одинаковым точкам считаются одной плоскостью, т.е. ABC, CBA, ACB и т.п. это одна и та же плоскость, в список включать только ABC). Вывести на экран названия и  уравнения плоскостей из списка и расстояние от  плоскостей до начала координат.

```python
list_planes = []
for point1, point2, point3 in combinations(points_list, 3):
  if not Point.is_collinear(point1, point2, point3):
    O = Point(0, 0, 0)
    plane = Plane(point1, point2, point3)
    list_planes.append(plane)
    name_plane = points_dict[point1] + points_dict[point2] + points_dict[point3]
    display(Latex(f'{latex(name_plane)}:\ уравнение\ плоскости\ {latex(plane.equation())};\ расстояние\ от\ плоскости\ до\ начала\ координат\ =\ {latex(plane.distance(O))}'))
```

###  Задание 2*
Составить список всех **различных** плоскостей, проходящих через любые три точки Задания 1. Вывести на экран названия и  уравнения каждой плоскости из списка и расстояние от  плоскости до самой далекой от нее точки.

```python
different_plane_list = list()
for point1, point2, point3 in combinations(points_list, 3):
  if not Point.is_collinear(point1, point2, point3):
    max_dist = 0
    plane = Plane(point1, point2, point3)
    check = True
    for i in range(len(different_plane_list)):
      if plane.equals(different_plane_list[i]):
        check = False
    if check:
      different_plane_list.append(plane)
      name_plane = points_dict[point1] + points_dict[point2] + points_dict[point3]
      for point in points_list:
        max_dist = max(plane.distance(point), max_dist)
      display(Latex(f'{latex(name_plane)}:\ уравнение\ плоскости\ {latex(plane.equation())};\ расстояние\ от\ плоскости\ до\ самой\ далекой\ от\ нее\ точки\ =\ {latex(max_dist)}'))
```

###  Задание 3
Точка $G(-1, 2, 0)$, остальные точки из Задания 1. Построить плоскость, перпендикулярную плоскости $ABG$, проходящую через $D$ и $F$. Построить плоскость, параллельную плоскости $ABF$, проходящую через $D$. Найти линию пересечения построенных плоскостей и расстояние от линии до $ABG$.

Вывести на экран уравнения построенных  плоскостей, линии пересечения $ABG$  и расстояние от линии  пересечения до $ABG$.

```python
G = Point(-1, 2, 0)
ABG = Plane(A, B, G)
ABF = Plane(A, B, F)
new_plane1 = ABG.perpendicular_plane(D, F)
new_plane2 = ABF.parallel_plane(D)
```

```python
display(Latex(f'Уровнение\ плоскости,\ перпендикулярной\ плоскости\ ABG:\ {latex(new_plane1.equation())}'))
```

```python
display(Latex(f'Уровнение\ плоскости,\ параллельной\ плоскости\ ABF:\ {latex(new_plane2.equation())}'))
```

```python
intersection_line = new_plane1.intersection(new_plane2)[0]
display(Latex(f'Уравнения\ линии\ пересечения\ построенных\ плоскостей:'))
for equation in intersection_line.equation():
  display(Eq(equation, 0))
```

```python
display(Latex(f'Расстояние\ от\ линии\ пересечения\ до\ ABG: {latex(ABG.distance(intersection_line))}'))
```
