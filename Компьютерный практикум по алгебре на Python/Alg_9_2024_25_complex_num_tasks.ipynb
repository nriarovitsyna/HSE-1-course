### Задание 1.
Даны комплексные числа вида
$$
z_k = k + \frac{i}{k},\quad k = 1, 2, 3.
$$
Вычислить (представить в алгебраической форме) с помощью sympy
$$
z_k \cdot z_{k + 1},\quad z_k^2,\quad \sqrt{z_k}.
$$

```python
numbers = [1, 2, 3]
for k in numbers:
  z1 = k + I / k
  z2 = (k + 1) + I / (k + 1)
  complex_numbers = [z1 * z2, z1 ** 2, z1 ** sympy.S.Half]
  display(*map(Latex, map(latex, map(sympy.expand_complex, complex_numbers))))
```

### Задание  3.
Изобразить на координатной плоскости точки, соответствующие корням  $3$-й степени комплексного числа $z = -\frac{1}{2} + i\frac{\sqrt{3}}{2}$. Подписать точки $A_k$, где $k$  - номер корня.

```python
n = 3
roots_list_3 = [sympy.root(-sympy.S(1) / 2 + sympy.S(3 ** sympy.S.Half) / 2 * I, n, k) for k in range(n)]
display(*map(Latex, map(latex, map(sympy.expand_complex, roots_list_3))))
for x, y, k_complex_z in zip(map(sympy.re, roots_list_3),
                              map(sympy.im, roots_list_3),
                              enumerate(roots_list_3)):
    plt.scatter(x, y)
    k, complex_z = k_complex_z
    plt.annotate(f'$A_{k}$',
            xy=(x, y),
            xytext=(x + 0.05, y - 0.05))
plt.axis('equal')
plt.xlabel("Re")
plt.ylabel("Im")
plt.xlim(-4, 4.5)
plt.ylim(-1.5, 1.5)
```

#### Задание  5*.
Изобразить корни 4 степени числа $5e^{\frac{i\pi}{8}}$, в полярной системе координат, подписи к точкам - аргументы (углы) сответствующих комплексных чисел.

Число вводить в тригонометрической форме, для получения корней комплексного числа $z = r(\cos(\varphi) + \sin(\varphi))$ использовать формулу
$\sqrt[n]{z} = \sqrt[n]{r}(\cos(\varphi_k) + i \sin(\varphi_k))$,
$\varphi_k = \frac{\varphi + 2\pi k}{n}$, $k = 0, ..., n$.

Вывести алгебраическую форму полученных корней на экран и сравнить с результатом, полученным с помощью sympy.root (с использованием sympy.expand_complex для упрощения выражения). Аналитическое представление оказывается разным.

Для каждого корня найти разность в формате float модуля разности корней, полученных двумя способами (т.е. нужно для каждой пары одинаковых корней, полученных двумя способами находить разность корней, вычислять модуль sympy.Abs этой разности и преобразовывать результат в формат float).

```python
from sympy import pi, sin, cos, root, Eq, symbols
n = 4
r = 5
r_n = r**(1/4)
arg = pi/8
z = r * (cos(arg) + I * sin(arg))
for k in range(n):
    arg_n = (arg + 2 * pi * k) / n
    root = r_n *(cos(arg_n) + I * sin(arg_n))
    sp_root = sympy.expand_complex(sympy.root(z, n, k))
    display(Latex(f'Корень,\ полученный\ с\ помощью\ формулы:'))
    display(root)
    display(Latex(f'Корень,\ полученный\ с\ помощью\ библиотеки\ sympy:'))
    display(sp_root)
    display(Latex(f'Разность\ корней:'))
    display(sympy.Abs(sp_root - root).evalf())
    plt.polar(arg_n, r_n, 'go')
    plt.annotate(arg_n, xy=(arg_n,r_n), xytext=(arg_n, r_n),
                  textcoords='offset points')
```

### Индивидуальное задание. Вариант №191
Дано уравнение $n$-й степени с переменной $z$. Вывести  уравнение и его точные корни  в алгебраической форме на экран, подписать корни  $q_1$, $q_2$, ...

Затем вывести на экран на отдельной строчке модули всех корней.

Найти и вывести на экран модуль и аргумент произведения всех корней.

```python
import sympy
z = sympy.Symbol('z')
equation = z ** 3 * (3 + 3 * I) + z ** 2 * (66 + 12 * I) + z * (324 - 144 * I) + 336 - 528 * I
display(Latex(f'Уравнение:\ {latex(equation.expand())}\ =\ 0'))
solution_root = sympy.roots(equation)
display(Latex(f'Корни\ в\ алгебраической\ форме:'))
display(Latex((r',\ '.join([rf'q_{k + 1} = {{{k}}}' for k\
in range(len(solution_root))])).format(*map(latex, map(sympy.expand_complex,
                                                      solution_root)))))
```

```python
moduli = list()
for root in solution_root:
  moduli.append(sympy.Abs(root))
display(Latex(f'Модули\ корней:'))
display(*moduli)
```

```python
product = sympy.prod(solution_root)
display(Latex(f'Модуль\ произведения\ всех\ корней:\ {latex(sympy.Abs(product))}'))
display(Latex(f'Аргумент\ произведения\ всех\ корней:\ {latex(sympy.functions.arg(product))}'))
```
