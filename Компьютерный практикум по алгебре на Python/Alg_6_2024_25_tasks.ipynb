### Задание 1.
Задать точки $A(1, 2, 3)$,
$B(1, 10, 3)$,
$C(-3, 2, 1)$,
$D(7, 2, 1)$ как матрицы-столбцы, вычислить и вывести на экран скалярное произведение векторов $AB$ и $CD$, их длины, косинус угла между векторами и угол в радианах и градусах.

```python
A = Matrix([1, 2, 3])
B = Matrix([1, 10, 3])
C = Matrix([-3, 2, 1])
D = Matrix([7, 2, 1])
ab = B - A
cd = C - D
```

```python
display(Latex(f'a \cdot b = {latex(ab.dot(cd))}'))
```

```python
length_ab = sympy.sqrt(ab[0] ** 2 + ab[1] ** 2 + ab[2] ** 2)
length_cd = sympy.sqrt(cd[0] ** 2 + cd[1] ** 2 + cd[2] ** 2)
display(Latex(f'Длина\ вектора\ ab = {length_ab}'))
display(Latex(f'Длина\ вектора\ cd = {length_cd}'))
```

```python
cos_phi = ab.dot(cd) / sympy.sqrt((ab.dot(ab)) * (cd.dot(cd)))
phi = sympy.acos(cos_phi)
display(Latex(fr"""\cos\phi = {latex(cos_phi)},\
\phi = {latex(phi)}\approx {round(phi, 4)},\
\phi = {phi * (180 / sympy.pi)}"""))
```

###  Задание 4.
Даны точки $P(-1, 2, -4)$ $Q(12, -2, 20)$ $R(0, 3, -6)$, $D(-16, 53, 11)$.

Определить, перпендикулярна ли плоскость $PQR$ прямой $DP$, $DQ$ или $DR$. Вывести на экран уравнения плоскости  $PQR$ и перпендикулярной ей прямой.

```python
P = Point(-1, 2, -4)
Q = Point(12, -2, 20)
R = Point(0, 3, -6)
D = Point(-16, 53, 11)
PQR = Plane(P, Q, R)
DP = Line(D, P)
DQ = Line(D, Q)
DR = Line(D, R)
```

```python
display(Latex(fr"""\text{{Уравнение плоскости PQR: }}{latex(PQR.equation())}"""))
```

```python
lines = [DP, DQ, DR]
name_lines = ["DP", "DQ", "DR"]
ans = list()
for i in range(len(lines)):
  if PQR.is_perpendicular(lines[i]):
    ans.append(name_lines[i])
for line in ans:
  display(Latex(fr"""\text{{Плоскость PQR перпендикулярна прямой }}{line}"""))
```

Даны точки в пространстве
$A$, $B$, $C$ и
$M$.

Найти угол между прямой $AB$ и плоскостью $CDM$, угол между $AC$ и $CM$, угол между плоскостями $ABC$ и $CDM$.

Составить уравнение:

a) плоскости, параллельной $ABC$ и проходящей через $M$,

b) плоскости, перпендикулярной $ABC$ и проходящей через $A$ и $M$,

c) прямой (параметрическое!), перпендикулярной $ABC$ и проходящей через $C$.

```python
A = Point3D(-3, 3, 5)
B = Point3D(-1, -2, -3)
C = Point3D(-3, 5, 5)
D = Point3D(3, 5, 3)
M = Point3D(-2, -2, 5)
AB = Line(A,B)
AC = Line(A, C)
CM = Line(C, M)
ABC_plane = Plane(A, B, C)
CDM_plane = Plane(C, D, M)
```

```python
angle_AB_CDM = CDM_plane.angle_between(AB)
angle_AC_CM = AC.angle_between(CM)
angle_ABC_CDM = ABC_plane.angle_between(CDM_plane)
display(Latex(fr"""\text{{Угол между прямой AB и плоскостью CDM: }}{latex(angle_AB_CDM)}"""))
display(Latex(fr"""\text{{Угол между векторами AC и CM: }}{latex(angle_AC_CM)}"""))
display(Latex(fr"""\text{{Угол между плоскостями ABC и CDM: }}{latex(angle_ABC_CDM)}"""))
```

```python
parallel_plane = ABC_plane.parallel_plane(M)
perpendicular_plane = ABC_plane.perpendicular_plane(A, M)
normal = ABC_plane.normal_vector
x, y, z, t = symbols('x y z t')
X = [x, y, z]
equations_line = [Eq(X[i], C[i] + normal[i] * t) for i in range(3)]
display(Latex(fr"""\text{{a) Уравнение плоскости, параллельной ABC и проходящей через M: }}{latex(parallel_plane.equation())}"""))
display(Latex(fr"""\text{{b) Уравнение плоскости, перпендикулярной ABC и проходящей через A и M: }}{latex(perpendicular_plane.equation())}"""))
display(Latex(fr"""\text{{c) Уравнение прямой:}}{latex(equations_line)}"""))
```
