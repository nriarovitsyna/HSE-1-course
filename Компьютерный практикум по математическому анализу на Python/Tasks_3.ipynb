### Задание 1
Решить уравнение $6x^4 + 9x^3 - 99x^2 - 66x + 360 = 0$ с помощью solve и solveset.

### Задание 6.
Найти решение уравнения $\frac{(x^3 + 5 x^2 - 3 x - 15)\ln(x^2 - 3)}{x^2 - 3 x + 2} = 0$. Выполнить проверку подстановкой subs.
#### Указание.
Решение нужно разбить на части. Нужно учесть область определения выражения в левой части уравнения.

```python
x = Symbol('x')
sol1 = solveset(x ** 3 + 5 * x ** 2 - 3 * x - 15)
sol2 = solveset(sympy.ln(x ** 2 - 3))
sol3 = solveset(x ** 2 - 3 * x + 2)
sol4 = Union(sol1, sol2)
sol5 = Union(sol4 - sol3)
display(sol5)
```

### Индивидуальное задание.
Через все точки пересечения графиков функций $f(x)$ и $g(x)$ провести горизонтальные прямые. Изобразить на графике $f(x)$ зеленым цветом, $g(x)$ красным, горизонтальные линии черным.
Формулы для $f(x)$ и $g(x)$ включить в легенду.

```python
from sympy import *
%matplotlib inline
x = Symbol('x')
res = solveset(111 * x ** 2 / 16 - 283 * x / 16 - 6 + (3 * x ** 2 - 4 * x - 3) / 2)
points = [111 * x ** 2 / 16 - 283 * x / 16 - 6 for x in res]
x_limits = (x, min(res) - 1, max(res) + 1)
p = plot(-(3 * x ** 2 - 4 * x - 3) / 2, x_limits, line_color='green', legend=True, label='$-((3x^2 - 4x - 3) / 2)$', show=False)
p.extend(plot(111 * x ** 2 / 16 - 283 * x / 16 - 6, x_limits, line_color='red', legend=True, label='$111x^2 / 16 - 283x / 16 - 6$', show=False))
for point in points:
  p.extend(plot(point, x_limits, line_color='black', show=False))
p.show()
```
