### Задача 2
Описать функцию, находящую неопределенный интеграл $\int f(x)\,dx$ в виде $F(x)+C$.

Использовать эту функцию для интегрирования $\sin(3x+5)$, $x{\rm tg}(x^2)$, $(e^{\sin x})\cos x$, $x^2e^{x^3}$

```python
def func(function_integrate):
  return integrate(function_integrate, x)
```

```python
x, C = symbols('x C')
func_ing = [sympy.sin(3 * x + 5), x * sympy.tan(x ** 2),sympy.exp(sympy.sin(x)) * sympy.cos(x), x ** 2 * sympy.exp(x ** 3)]
for i in func_ing:
  display(func(i) + C)
```

### Задача 3.
Найти первообразную функции $\frac{x-3}{(x+1)^2}$, график которой проходит через начало координат.

```python
x = symbols('x')
f = (x - 3) / (x + 1) ** 2
F = integrate(f, x)
C = -F.subs(x, 0)
F = F + C
display(F)
```

### Задача 4.
Описать функцию, вычисляющую неопределенный интеграл от $\frac{1}{\sqrt{x^2+a}}$ по формуле "длинного логарифма":
$$
\int\frac{1}{\sqrt{x^2+a}} = \ln|x + \sqrt{x^2 + a}| + C,\quad a \neq 0.
$$
Функция возвращает $\ln|x| + C$ при $a = 0$. Аргументы функции - переменная интегрирования $x$ и параметр $a$.

Вызвать функцию при $a = 0, 1, 4$. Сравнить с результатом integrate.

Проверить дифференцированием, при необходимости использовать simplyfy для того, чтобы упростить выражение.

Рекомендется создавать положительные символы, чтобы не было неоднозначности.

```python
def integ_log(x,a):
  C = symbols('C')
  if a != 0:
    return sympy.log(abs(x + sympy.sqrt(x ** 2 + a))) + C
  else:
    return sympy.log(abs(x)) + C
```

```python
x, a = symbols('x a')
numb = [0, 1, 4]
function = 1 / sympy.sqrt(x ** 2 + abs(a))
for i in numb:
  print(f"Значение функции для a = {i}:")
  display(integ_log(x, i))
  print(f"Значение функции для a = {i} с использованием integrate:")
  res = function.subs(a, i)
  display(integrate(res, x))
```

### Задача 5.
Провести замену переменной $y = x^2$ в интеграле $\int x e^{x^2}\,dx$.

```python
y = Symbol('y', positive=True)
x = symbols('x')
function_5 = Integral(x * sympy.E ** (x ** 2))
sympy.Eq(function_5, function_5.transform(x ** 2, y))
```

### Индивидуальное задание. Вариант №191
Провести замену переменных в определенном интеграле, представить результат в виде двух уравнений, как в примере:
\begin{align*}
\int_1^4 \frac{dx}{\sqrt{x}\left(\sqrt{x}+1\right)} = \int_1^2 \frac{2\,dy}{y + 1}\\
\int_1^2 \frac{2\,dy}{y + 1} = 2(\ln 3 - \ln 2)
\end{align*}
(с точностью до перестановки слагаемых и сомножителей)

```python
import sympy as sp

original_integral = sp.E ** x / (sp.E ** (2 * x) + 1)
lower_limit = sp.log(sp.S(sp.sqrt(3)) / 3)
upper_limit = sp.log(sp.S(sp.sqrt(3)))

integral = Integral(original_integral, (x, lower_limit, upper_limit))


integral_subs = integral.transform(sp.E ** x, y)

integral_result = sp.integrate(original_integral,(x, lower_limit, upper_limit))

display(sympy.Eq(integral, integral_subs))
display(sympy.Eq(integral_subs, integral_result))
```
