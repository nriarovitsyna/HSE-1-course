### Задание 1
Найти точки экстремума:

1) $z = x^2 + xy + y^2 - 2x - y$

2) $z = x^3y^2(6 - x - y)$, $(x, y > 0)$

3) $z = \frac{8}{x} + \frac{x}{y} + y$

4) $z = \frac{1 + x - y}{\sqrt{1 + x^2 + y^2}}$
###### Указание.
Вначале найти стационарные точки. Составить определитель из вторых производных в произвольной точке. Подставляя найденные (вещественные) стационарные точки, решить вопрос о наличии экстремума.

```python
from sympy.abc import x, y
```

```python
f = x ** 2 + x * y + y ** 2 - 2 * x - y

stat_points = nonlinsolve([f.diff(x), f.diff(y)], [x, y])
display(*stat_points)

H = Matrix([[f.diff(x, 2), f.diff(x, y)], [f.diff(x, y), f.diff(y, 2)]])

for point in stat_points:

    x0, y0 = point
    determinant = H.subs({x: x0, y: y0}).det()

    if determinant > 0:
        if f.diff(x, 2).subs({x: x0, y: y0}) > 0 and f.diff(y, 2).subs({x: x0, y: y0}) > 0:
            display(f'Минимум: ({x0}, {y0})')
        elif f.diff(x, 2).subs({x: x0, y: y0}) < 0 and f.diff(y, 2).subs({x: x0, y: y0}) < 0:
            display(f'Максимум: ({x0}, {y0})')
```

```python
f = x ** 3 * y ** 2 * (6 - x - y)

stat_points = nonlinsolve([f.diff(x), f.diff(y)], [x, y])
display(*stat_points)
```

```python
f = 8 / x + x / y + y

stat_points = nonlinsolve([f.diff(x), f.diff(y)], [x, y])
display(*stat_points)

H = Matrix([[f.diff(x, 2), f.diff(x, y)], [f.diff(x, y), f.diff(y, 2)]])

for point in stat_points:
    x0, y0 = point
    if x0 != 0 and y0 != 0:
        determinant = H.subs({x: x0, y: y0}).det()
        if determinant.is_real:
            if determinant > 0:
                A = f.diff(x, 2).subs({x: x0, y: y0})
                B = f.diff(y, 2).subs({x: x0, y: y0})
                if A > 0 or B > 0:
                    display(f'минимум: {point}')
                elif A < 0 or B < 0:
                    display(f'Максимум: {point}')
                else:
                    display(f'нет экстремума: {point}')
            elif determinant < 0:
                display(f'нет экстремума:', point)
            else:
                display(f'требуется дальнейшее исследование: {point}')
        else:
            display(f'точка не удовлетворяет условию x, y != 0: {point}')
    else:
        display(f'точка не удовлетворяет условию x, y != 0: {point}')
```

### Задание 2.
С помощью метода Лагранжа найти условный экстремум функции
$z = x + 2y$ при условии $x^2 + y^2 = 5$.

```python
x, y, lam = symbols('x y lamda')
varrs = [x, y, lam]
z = x + 2 * y
restriction = x ** 2 + y ** 2 - 5
L = z + lam * restriction
stat_points = nonlinsolve([L.diff(var)for var in varrs], varrs)
display(stat_points)
```

```python
dx, dy, dx2, dy2 = symbols('dx dy dx^2 dy^2')
d2L = L.diff(x, 2) * dx2 + 2 * L.diff(x, y) * dx * dy + L.diff(y, 2) * dy2
factor(d2L)
```

```python
zz = Symbol('z')
for point in stat_points:
    display(point[:-1], sp.Eq(zz, z.subs({var: point[j] for j, var in enumerate(varrs)})))
```

### Задание 4.
С помощью метода Лагранжа найти экстремум функции $u = x - 2y + 2z$ при условии $x^2 + y^2 + z^2 = 9$.

```python
x, y, z, lam = symbols('x y z lambda')
varrs = [x, y, z, lam]
u = x - 2 * y + 2 * z
restriction = x ** 2 + y ** 2 + z ** 2 - 9
L = u + lam * restriction
stat_points = nonlinsolve([L.diff(var)for var in varrs], varrs)
display(stat_points)
```

```python
dx, dy, dz, dx2, dy2, dz2 = symbols('dx dy dz dx^2 dy^2 dz^2')
d2L = 0
diffs = {x: (dx, dx2), y: (dy, dy2), z: (dz, dz2)}
for i, var in enumerate(varrs[:-1]):
    d2L += L.diff(var, 2)*diffs[var][1]
    for j in range(i + 1, 3):
        d2L += 2 * L.diff(var, varrs[j]) * diffs[var][0] * diffs[varrs[j]][0]
```
