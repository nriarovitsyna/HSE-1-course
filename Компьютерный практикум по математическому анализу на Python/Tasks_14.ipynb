### Задача 2.
Построить на отдельных графиках линии уровня функций $z=\ln(x^2+y)$ и $z=x^2-y^2$.

```python
x, y = symbols('x y')
C = symbols('C', real=True)
```

```python
def contourline(f, x, y, C):
    return sympy.Eq(f(x, y), C)
```

```python
def f(x, y):
    return Log(x ** 2 + y)
```

```python
p = plot(ylabel='y', show=False)
for item in [(-2, 'blue'), (-1, 'green'), (0, 'red'), (2, 'black'), (5, 'pink')]:
    c, col = item
    p.append(plot_implicit(contourline(f, x, y, c), line_color=col, line_width=15, show=False)[0])
p.show()
```

### Задача 6.
Проверить выполнение теоремы Эйлера для функции $f(x,y)=\frac{x}{x^2+y^2} $.
###### Указание.
Упростить выражение с помощью simplify, прежде чем сравнивать.

```python
from sympy import diff, ln, simplify
```

```python
x, y = symbols('x y')

f = simplify(x / (x ** 2 + y ** 2))

df_dx = diff(f, x)
df_dy = diff(f, y)

left_side = simplify(x * df_dx + y * df_dy)
right_side = f

print("Левая часть уравнения теоремы Эйлера:")
print(left_side)
```

```python
print("Правая часть уравнения теоремы Эйлера:")
print(right_side)
```

```python
print("Сравнение:")
print(left_side + right_side == 0)
```
