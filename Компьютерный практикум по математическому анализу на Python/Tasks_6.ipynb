$$
y = x^2e^{-x}
$$

```python
from sympy.abc import x
def f(x):
  return x ** 2 * sympy.exp(-x)
D = continuous_domain(f(x), x, S.Reals)
D
```

```python
if f(-x) == f(x):
    print('Четная функция')
elif f(-x) == -f(x):
    print('Нечетная функция')
else:
    print('Функция общего вида')
```

```python
roots = list(solveset(f(x), domain=S.Reals))
roots.sort()
roots
```

```python
def df(x):
    return simplify(f(x).diff(x))
display(df(x))
```

```python
critical_points = sympy.solve(df(x))
critical_points
```

```python
display(df(x).subs(x, -2), df(x).subs(x, 1), df(x).subs(x, 4))
```

```python
def df2(x):
    return simplify(f(x).diff(x, 2))
display(df2(x))
print("Точки перегиба:")
inflection_points = sympy.solve(df2(x))
display(*inflection_points)
```

```python
display(df2(x).subs(x, 0), df2(x).subs(x, 2), df2(x).subs(x, 5))
```

```python
k = sympy.limit(f(x)/x, x, +sympy.oo)
k
```

```python
k = sympy.limit(f(x)/x, x, -sympy.oo)
k
```

```python
y = x**2 * sympy.exp(-x)
y_prime = sympy.diff(y, x)
f_prime = y.diff(x)
f_double_prime = f_prime.diff(x)
critical_points = sympy.solve(f_prime, x)
inflection_points = sympy.solve(f_double_prime, x)
print("Точки экстремума:")
display(*critical_points)
print("Точки перегиба:")
display(*[i.evalf() for i in inflection_points])
ax = plt.gca()
x_vals = np.linspace(-1, 4, 400)
y_vals = np.array([float(y.subs(x, val)) for val in x_vals])
ax.plot(x_vals, y_vals, label='y=x^2e^(-x)')
ax.set_ylim(-1, 1)
ax.axhline(0, color='r', linestyle='--', label='y=0 Гориз. асимптота')
number = [r'$A$', r'$C$']
number1 = [r'$B$', r'$D$']
for i,point in enumerate(critical_points):
    label = f'Point {number[i]}: ({point:.2f}, {f(point):.2f})'
    ax.scatter(point, y.subs(x, point), color='m', marker='o', label='Экстремум')
    ax.text(point, f(point), label, fontsize=12, ha='right', va='bottom')
for i,point in enumerate(inflection_points):
    label = f'Point {number1[i]}: ({point:.2f}, {f(point):.2f})'
    ax.scatter(point, y.subs(x, point),color='b', marker='o', label='Точка перегиба')
    ax.text(point, f(point), label, fontsize=12, ha='right', va='bottom')
ax.text(0.5, 0, f'Горизонтальная асимптота y=0 ', fontsize=8)
ax.legend(loc='lower right')
plt.title('График функции y=$x^2*e^{-x}$')
ax.spines['right'].set_color('none')
ax.spines['top'].set_color('none')
ax.spines['bottom'].set_position(('data', 0))
ax.spines['left'].set_position(('data', 0))
```

## Задание 2
Исследовать функцию с помощью производной, найти необходимые пределы и решить уравнения.

Построить график функции и асимптот, отметить и подписать точки экстремума и точки перегиба,
включить функцию и асимптоты в легенду.
$$
y=xe^{2/x} + 1
$$

```python
def f(x):
    return x * sympy.exp(2 / x) + 1
x = sympy.symbols('x')
D = continuous_domain(f(x), x, S.Reals)
D
```

```python
if f(-x) == f(x):
    print('Четная функция')
elif f(-x) == -f(x):
    print('Нечетная функция')
else:
    print('Функция общего вида')
```

```python
roots = []
for i in range(-1, 1):
    if i == 0:
        continue
    root = nsolve(f(x), x, i)
    roots.append(root)

roots = list(set(roots))
roots.sort()
print(roots)
```

```python
def df(x):
    return simplify(f(x).diff(x))
display(df(x))
critical_points = sympy.solve(df(x))
print("Точки экстремума:")
critical_points
```

```python
display(df(x).subs(x, 1), df(x).subs(x, 2.5))
```

```python
def df2(x):
    return simplify(f(x).diff(x, 2))
display(df2(x))
print("Точки перегиба:")
inflection_points = sympy.solve(df2(x))
display(inflection_points)
```

```python
horizontal_asymptote_plus_inf = sympy.limit(f(x),x, sympy.oo)
horizontal_asymptote_min_inf = sympy.limit(f(x), x, -sympy.oo)
display(horizontal_asymptote_plus_inf)
display(horizontal_asymptote_min_inf )
k = sympy.limit(f(x)/x, x, sympy.oo)
display(k)
b_plus_inf = sympy.limit(f(x) - k*x, x, sympy.oo)
display(b_plus_inf)
b_min_inf = sympy.limit(f(x) - k*x, x, -sympy.oo)
display(b_min_inf)
```

```python
print(sympy.limit(f(x),x, 0,dir='-'),sympy.limit(f(x),x, 0))
```
